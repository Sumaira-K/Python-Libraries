### Merge, Join, and Concat

```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Ali", "Sara", "John"]
})

print(students)
```

```python
marks = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Marks": [85, 90, 78]
})

print(marks)
```

```python
result = pd.merge(students, marks, on="Student_ID")

print(result)
```

```python
pd.merge(students, marks, on="Student_ID")
```

```python
pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)
```

```python
pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)
```

```python
pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)
```

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"]
})

df2 = pd.DataFrame({
    "Name": ["John", "Emma"]
})

pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```

### Pivot Tables (pivot_table())

```python
import pandas as pd

data = {
    "City": ["Delhi", "Delhi", "Mumbai", "Mumbai", "Delhi"],
    "Product": ["Laptop", "Mobile", "Laptop", "Mobile", "Mobile"],
    "Sales": [50000, 20000, 55000, 18000, 15000]
}

df = pd.DataFrame(data)

print(df)
```

```python
pd.pivot_table(
    df,
    index="City",
    values="Sales"
)
```

```python
pd.pivot_table(
    df,
    index="City",
    values="Sales",
    aggfunc="sum"
)
```

```python
pd.pivot_table(
    df,
    index="City",
    values="Sales",
    aggfunc="max"
)
```

```python
pd.pivot_table(
    df,
    index="City",
    values="Sales",
    aggfunc=["sum", "mean", "max"]
)
```

```python
pd.pivot_table(
    df,
    index="City",
    columns="Product",
    values="Sales",
    aggfunc="sum"
)
```

### pd.crosstab()
