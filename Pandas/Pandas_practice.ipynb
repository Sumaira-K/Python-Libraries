### Filtering Data in Pandas

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Emma", "David"],
    "Age": [20, 19, 21, 22, 18],
    "Marks": [85, 90, 78, 95, 65],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df[df["Marks"] > 80]
```

```python
df[df["Age"] < 21]
```

```python
df[df["City"] == "Delhi"]
```

```python
df[df["City"] != "Mumbai"]
```

```python
df[(df["Marks"] > 80) & (df["Age"] > 20)]
```

```python
df[(df["City"] == "Delhi") | (df["Marks"] > 90)]
```

```python
df[df["City"].isin(["Delhi", "Pune"])]
```

```python
df[df["Age"].between(19, 21)]
```

```python
df[df["Marks"] > 80]["Name"]
```

```python
df[df["Marks"] > 80][["Name", "Marks"]]
```

### GroupBy

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Emma", "David"],
    "City": ["Delhi", "Delhi", "Mumbai", "Mumbai", "Pune"],
    "Marks": [85, 90, 78, 95, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.groupby("City")
```

```python
df.groupby("City")["Marks"].mean()
```

```python
df.groupby("City")["Marks"].sum()
```

```python
df.groupby("City")["Name"].count()
```

```python
df.groupby("City")["Marks"].max()
```

```python
df.groupby("City")["Marks"].min()
```

```python
df.groupby("City")["Marks"].agg(["mean", "sum", "max", "min", "count"])
```

```python
df.groupby("City")["Marks"].mean().reset_index()
```

### Merge, Join, and Concat

```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Ali", "Sara", "John"]
})

print(students)
```

```python
marks = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Marks": [85, 90, 78]
})

print(marks)
```

```python
result = pd.merge(students, marks, on="Student_ID")

print(result)
```

```python
pd.merge(students, marks, on="Student_ID")
```

```python
pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)
```

```python
pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)
```

```python
pd.merge(
    students,
    marks,
    on="Student_ID",
    how="outer"
)
```

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"]
})

df2 = pd.DataFrame({
    "Name": ["John", "Emma"]
})

pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```
