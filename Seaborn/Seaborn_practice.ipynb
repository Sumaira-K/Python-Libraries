## Line Plot

```python
sns.lineplot(
    data=tips,
    x="size",
    y="total_bill"
)

plt.show()
```

## Scatter Plot

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Bar Plot

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
```

## Count Plot

```python
sns.countplot(
    data=tips,
    x="day"
)

plt.show()
```

## Histogram

```python
sns.histplot(
    data=tips,
    x="total_bill"
)

plt.show()
```

## KDE Plot

```python
# Kernal DEnsity Plot
sns.kdeplot(
    data=tips,
    x="total_bill"
)

plt.show()
```

## Box Plot

```python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
```

## Violin Plot

```python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
```

## Pair Plot

```python
sns.pairplot(tips)

plt.show()
```

## Heatmap

```python
corr = tips.corr(numeric_only=True)
```

```python
sns.heatmap(corr)

plt.show()
```

## Improved Heatmap

```python
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)
```

## Saving the Graph

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

plt.figure(figsize=(8, 5))

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.title("Tips vs Total Bill")

plt.tight_layout()

plt.savefig("graph.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.title("Tips vs Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```
