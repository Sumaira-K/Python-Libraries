## Boolean Masking

```python
marks = np.array([35, 80, 55, 20, 90])
print(marks > 50)
```

```python
print(marks[marks > 50])
```

```python
ages = np.array([12, 18, 25, 16, 30])
adults = ages >= 18

print(adults)
```

```python
print(ages[adults])
```

```python
numbers = np.array([1,2,3,4,5,6,7,8])

print(numbers[numbers % 2 == 0])
```
