## Animations

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
```

```python
x = []
y = []

fig, ax = plt.subplots()

def update(frame):
    x.append(frame)
    y.append(frame**2)

    ax.clear()
    ax.plot(x, y)

ani = FuncAnimation(fig, update, frames=10, interval=500)

plt.show()
```

## Twin Axis

```python
import matplotlib.pyplot as plt

months = [1,2,3,4]

sales = [100,150,200,250]

temperature = [20,25,28,30]

fig, ax1 = plt.subplots()

ax1.plot(months, sales)

ax2 = ax1.twinx()

ax2.plot(months, temperature, linestyle="--")

plt.show()
```

## Interactive Widgets

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

x = np.linspace(0,10,100)

fig, ax = plt.subplots()

plt.subplots_adjust(bottom=0.25)

a = 1

line, = plt.plot(x, a*x)

slider_ax = plt.axes([0.2,0.1,0.6,0.03])

slider = Slider(slider_ax,"Slope",0,5,valinit=1)

def update(val):
    line.set_ydata(slider.val*x)
    fig.canvas.draw_idle()

slider.on_changed(update)

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [78, 85, 92, 88]

plt.figure(figsize=(8, 5))

plt.plot(students, marks, marker="o", linewidth=2)

plt.title("Student Marks", fontsize=16)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()

plt.savefig("student_marks.png", dpi=300)

plt.show()
```
