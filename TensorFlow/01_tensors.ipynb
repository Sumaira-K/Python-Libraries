### GradientTape (Automatic Differentiation)

```python
import tensorflow as tf

x = tf.Variable(3.0)

with tf.GradientTape() as tape:
    y = x ** 2

gradient = tape.gradient(y, x)

print(gradient)
```

### Gradient Descent Algorithm

```python
import tensorflow as tf

w = tf.Variable(10.0)
learning_rate = 0.1

for step in range(10):

    with tf.GradientTape() as tape:
        loss = (w - 5) ** 2

    gradient = tape.gradient(loss, w)

    w.assign_sub(learning_rate * gradient)

    print("Step:", step + 1, "Weight:", w.numpy(), "Loss:", loss.numpy())
```

### Keras Optimizers

```python
import tensorflow as tf

w = tf.Variable(10.0)

optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)

for step in range(10):

    with tf.GradientTape() as tape:
        loss = (w - 5) ** 2

    gradient = tape.gradient(loss, w)

    optimizer.apply_gradients([(gradient, w)])

    print("Step:", step + 1, "Weight:", w.numpy(), "Loss:", loss.numpy())
```

### Implementing a neuron

```python
import tensorflow as tf

# Inputs
inputs = tf.constant([4.0, 2.0])

# Weights
weights = tf.constant([0.5, 0.3])

# Bias
bias = tf.constant(1.0)

# Weighted sum
z = tf.reduce_sum(inputs * weights) + bias

# Activation functions
relu_output = tf.keras.activations.relu(z)
sigmoid_output = tf.keras.activations.sigmoid(z)

print("Weighted sum:", z.numpy())
print("ReLU:", relu_output.numpy())
print("Sigmoid:", sigmoid_output.numpy())
```

```python
import tensorflow as tf

# 1. Create inputs
inputs = tf.constant([3.0, 4.0])

# 2. Create weights
weights = tf.Variable([0.5, 0.8])

# 3. Create bias
bias = tf.Variable(-5.0)

# 4. Calculate weighted sum
z = tf.reduce_sum(inputs * weights) + bias

# 5. Apply ReLU
output = tf.keras.activations.relu(z)

print("Weighted sum:", z.numpy())
print("Neuron output:", output.numpy())
```

```python
import tensorflow as tf

X = tf.constant([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 2.0],
    [4.0, 3.0],
    [5.0, 1.0]
])

y = tf.constant([
    [8.0],
    [7.0],
    [12.0],
    [17.0],
    [13.0]
])
```

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)
```

```python
history = model.fit(X, y, epochs=100, verbose=1)
```

```python
new_data = tf.constant([[6.0, 2.0]])

prediction = model.predict(new_data)

print(prediction)
```

```python
predictions = model.predict(X)

print("Actual values:")
print(y.numpy())

print("\nPredicted values:")
print(predictions)
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss="mse",
    metrics=["mae"]
)

history = model.fit(
    X,
    y,
    epochs=1000,
    verbose=0
)
```

```python
predictions = model.predict(X)

print("Actual values:")
print(y.numpy())

print("\nNew predictions:")
print(predictions)

print("\nFinal loss:")
print(history.history["loss"][-1])
```

```python
new_data = tf.constant([
    [6.0, 2.0],
    [7.0, 3.0],
    [10.0, 5.0]
])

predictions = model.predict(new_data)

print(predictions)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X.numpy(),
    y.numpy(),
    test_size=0.2,
    random_state=42
)

print("Training inputs:", X_train.shape)
print("Testing inputs:", X_test.shape)

print("Training outputs:", y_train.shape)
print("Testing outputs:", y_test.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss="mse",
    metrics=["mae"]
)

history = model.fit(
    X_train,
    y_train,
    epochs=1000,
    verbose=0
)
```

```python
test_loss, test_mae = model.evaluate(X_test, y_test)

print("Testing Loss:", test_loss)
print("Testing MAE:", test_mae)
```

```python
test_prediction = model.predict(X_test)

print("Testing input:")
print(X_test)

print("\nActual output:")
print(y_test)

print("\nPredicted output:")
print(test_prediction)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=1000,
    validation_split=0.2,
    verbose=0
)
```

```python
print("Training loss:", history.history["loss"][-1])

print("Validation loss:", history.history["val_loss"][-1])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training Loss vs Validation Loss")

plt.legend()
plt.grid(True)
plt.show()
```

```python
# Early stopping
model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss="mse",
    metrics=["mae"]
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=1000,
    validation_split=0.2,
    callbacks=[early_stopping],
    verbose=1
)
```
